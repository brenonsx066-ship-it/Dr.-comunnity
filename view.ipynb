```python
from datetime import datetime

def exibir_cabecalho(titulo):
    print("\n" + "=" * 65)
    print(titulo.center(65))
    print("=" * 65)


def pausar():
    input("\nPressione ENTER para continuar...")


def obter_data():
    return datetime.now().strftime("%d/%m/%Y %H:%M")


def validar_data(data):
    try:
        datetime.strptime(data, "%d/%m/%Y")
        return True
    except ValueError:
        return False


def validar_horario(horario):
    try:
        datetime.strptime(horario, "%H:%M")
        return True
    except ValueError:
        return False


def mostrar_medicos(medicos):
    if not medicos:
        print("\nNenhum médico encontrado.")
        return

    for medico in medicos:
        medico_id, nome, especialidade = medico

        print(f"\nID: {medico_id}")
        print(f"Médico: {nome}")
        print(f"Especialidade: {especialidade}")
        print("-" * 40)
```
